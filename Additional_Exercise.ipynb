```python
# Student Inputs
last_name = "DeJesus"
seed_num = 3
favorite_artist = "BEN&BEN"

sources = [f"{last_name}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Generate values using seed_num and character positions of BEN&BEN (e.g., B=2, E=5, N=14...)
artist_positions = [ord(char.upper()) - 64 for char in favorite_artist if char.isalpha()]
values = [seed_num * pos for pos in artist_positions[:5]]

test_data = [
    (sources[0], categories[0], values[0]),
    (sources[1], categories[1], values[1]),
    (sources[2], categories[2], values[2]),
    (sources[0], categories[0], values[0]),  # Repeated record
    (sources[3], categories[3], values[3])
]

print("=== 1. Generated Test Data ===")
for record in test_data:
    print(record)

# 2. Organize Test Data by Source and Category
organized_data = {}
for source, category, value in test_data:
    if source not in organized_data:
        organized_data[source] = {}
    if category not in organized_data[source]:
        organized_data[source][category] = []
    organized_data[source][category].append(value)

print("\n=== 2. Organized Test Data ===")
for source, cats in organized_data.items():
    print(f"Source: {source} -> {cats}")

# 3. Identify Repeated and Distinct Data
unique_records = set(test_data)
repeated_records = [rec for rec in test_data if test_data.count(rec) > 1]
print("\n=== 3. Repeated & Distinct Data ===")
print("Unique Count:", len(unique_records))
print("Repeated Records Found:", list(set(repeated_records)))

# 4. Numerical and Categorical Summaries
all_values = [val for _, _, val in test_data]
print("\n=== 4. Numerical and Categorical Summaries ===")
print("Min Value:", min(all_values))
print("Max Value:", max(all_values))
print("Average Value:", sum(all_values) / len(all_values))

# 5. Processed Results
processed_results = [(src, cat, sum(vals)) for src, cats in organized_data.items() for cat, vals in cats.items()]
print("\n=== 5. Processed Results ===")
for res in processed_results:
    print(res)

# 6. Ordered Results (Sorted by Source)
ordered_results = sorted(processed_results, key=lambda x: x[0])
print("\n=== 6. Ordered Results ===")
for res in ordered_results:
    print(res)

# 7. Final Consolidated Summary Report
print("\n=== 7. Final Engineering Report ===")
print(f"Total Unique Records Analyzed: {len(unique_records)}")
print("Engineering data analysis finished!")
```

```python
# Student Inputs
last_name = "DeJesus"
seed_num = 3
favorite_artist = "BEN&BEN"

student_name = f"Student_{last_name}"
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

# Generate scores using seed_num and character positions of BEN&BEN (e.g., B=2, E=5, N=14...)
artist_positions = [ord(char.upper()) - 64 for char in favorite_artist if char.isalpha()]
scores = [seed_num * pos for pos in artist_positions[:5]]

# 1. Generated Performance Records (Name, Subject, Score)
performance_records = [
    (student_name, subjects[0], scores[0]),
    (student_name, subjects[1], scores[1]),
    (student_name, subjects[2], scores[2]),
    (student_name, subjects[0], scores[0]),  # Repeated record
    (student_name, subjects[3], None)           # Incomplete record (missing score)
]

print("=== 1. Generated Performance Records ===")
for rec in performance_records:
    print(rec)

# 2. Organized Student/Subject Data
organized_data = {}
for name, subj, score in performance_records:
    if name not in organized_data:
        organized_data[name] = []
    organized_data[name].append((subj, score))

print("\n=== 2. Organized Student Data ===")
for name, items in organized_data.items():
    print(f"{name}: {items}")

# 3. Repeated and Incomplete Records
repeated = []
incomplete = []

for rec in performance_records:
    if performance_records.count(rec) > 1 and rec not in repeated:
        repeated.append(rec)
    if rec[2] is None:
        incomplete.append(rec)

print("\n=== 3. Repeated / Incomplete Records ===")
print("Repeated Records:", repeated)
print("Incomplete Records:", incomplete)

# 4. Performance Categories (Filter out None values)
valid_scores = []
for _, _, score in performance_records:
    if score is not None:
        valid_scores.append(score)

print("\n=== 4. Performance Categories (Valid Scores) ===")
print("Scores:", valid_scores)

# 5. Student and Subject Summaries
print("\n=== 5. Student and Subject Summaries ===")
print("Highest Score:", max(valid_scores))
print("Lowest Score:", min(valid_scores))
print("Average Score:", sum(valid_scores) / len(valid_scores))

# 6. Ordered Results (Sorted by Subject name)
valid_records = [r for r in performance_records if r[2] is not None]
valid_records.sort(key=lambda x: x[1])

print("\n=== 6. Ordered Results (Sorted by Subject) ===")
for rec in valid_records:
    print(rec)

# 7. Final Consolidated Student Performance Report
print("\n=== 7. Final Consolidated Student Performance Report ===")
print("--------------------------------------------------")
print(f"Student Name : {student_name}")
print(f"Total Subjects: {len(subjects)}")
print(f"Valid Scores Analyzed: {len(valid_scores)}")
print(f"Highest Score: {max(valid_scores)}")
print(f"Lowest Score: {min(valid_scores)}")
print(f"Class Average: {round(sum(valid_scores) / len(valid_scores), 2)}")
print("Status: Passed / Completed")
print("--------------------------------------------------")
```

```python
# Student Inputs
last_name = "DeJesus"
seed_num = 3
favorite_artist = "BEN&BEN"

# 1. Generate Transaction Data
customers = [f"{last_name}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {
    products[0]: 50, 
    products[1]: 20, 
    products[2]: 350, 
    products[3]: 500, 
    products[4]: 1200
}

# Calculate quantities using seed_num and character positions of BEN&BEN (e.g., B=2, E=5, N=14...)
artist_positions = [ord(char.upper()) - 64 for char in favorite_artist if char.isalpha()]
quantities = [seed_num * pos for pos in artist_positions[:5]]

# Transaction Records: (Customer, Product, Quantity)
transaction_records = [
    (customers[0], products[0], quantities[0]),
    (customers[1], products[1], quantities[1]),
    (customers[2], products[2], quantities[2]),
    (customers[0], products[0], quantities[0]),  # Repeated purchase
    (customers[3], products[3], quantities[3])
]

print("=== 1. Generated Transaction Records ===")
for rec in transaction_records:
    print(rec)

# 2. Organize Transaction Data
organized_trans = {}
for cust, prod, qty in transaction_records:
    if cust not in organized_trans:
        organized_trans[cust] = []
    organized_trans[cust].append((prod, qty))

print("\n=== 2. Organized Transaction Data ===")
for cust, items in organized_trans.items():
    print(f"{cust}: {items}")

# 3. Repeated and Distinct Transactions
unique_trans = set(transaction_records)
repeated_trans = [r for r in transaction_records if transaction_records.count(r) > 1]
print("\n=== 3. Repeated / Distinct Transactions ===")
print("Unique Transactions Count:", len(unique_trans))
print("Repeated Purchases:", list(set(repeated_trans)))

# 4. Transaction Values (Calculating total per record)
print("\n=== 4. Transaction Values ===")
trans_values = []
for cust, prod, qty in transaction_records:
    total_price = qty * prices[prod]
    trans_values.append((cust, prod, qty, total_price))
    print(f"{cust} bought {qty} {prod} -> Total: ₱{total_price}")

# 5. Customer and Product Summaries
print("\n=== 5. Customer and Product Summaries ===")
total_store_sales = sum([val[3] for val in trans_values])
print(f"Overall Store Sales: ₱{total_store_sales}")

# 6. Ordered Results (Sorted by Customer name)
ordered_trans = sorted(trans_values, key=lambda x: x[0])
print("\n=== 6. Ordered Results ===")
for item in ordered_trans:
    print(item)

# 7. Final Transaction Report
print("\n=== 7. Final Transaction Report ===")
print(f"Total Transactions Analyzed: {len(transaction_records)}")
print("Transaction analysis finished!")
```
